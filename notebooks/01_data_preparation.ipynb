# 01 — Data Preparation

This notebook prepares anonymized questionnaire and controlled-task data.

**Important:** real participant data are not included in the repository.  
Copy the CSV templates in `data/raw/`, rename the working files to:

- `questionnaire.csv`
- `controlled_task.csv`

Then add anonymized data locally.


```python
from pathlib import Path
import sys
import pandas as pd

ROOT = Path.cwd().resolve()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent

sys.path.insert(0, str(ROOT))

from src.data_cleaning import clean_questionnaire, clean_controlled_task

RAW = ROOT / "data" / "raw"
PROCESSED = ROOT / "data" / "processed"
PROCESSED.mkdir(parents=True, exist_ok=True)

print("Project root:", ROOT)
```

```python
questionnaire_path = RAW / "questionnaire.csv"

if questionnaire_path.exists():
    questionnaire = pd.read_csv(questionnaire_path)
    questionnaire_clean = clean_questionnaire(questionnaire)
    display(questionnaire_clean.head())
    questionnaire_clean.to_csv(PROCESSED / "questionnaire_clean.csv", index=False)
    print("Saved questionnaire_clean.csv")
else:
    print("No questionnaire.csv found.")
    print("Use questionnaire_template.csv as the column template.")
```

```python
controlled_path = RAW / "controlled_task.csv"

if controlled_path.exists():
    controlled = pd.read_csv(controlled_path)
    controlled_clean = clean_controlled_task(controlled)
    display(controlled_clean.head())
    controlled_clean.to_csv(PROCESSED / "controlled_task_clean.csv", index=False)
    print("Saved controlled_task_clean.csv")
else:
    print("No controlled_task.csv found.")
    print("Use controlled_task_template.csv as the column template.")
```
