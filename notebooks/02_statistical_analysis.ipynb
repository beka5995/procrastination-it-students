# 02 — Statistical Analysis

This notebook follows the SS5 analysis plan.

- **RQ1:** descriptive statistics + Friedman test
- **RQ2 survey:** Spearman correlation
- **RQ2 controlled task:** paired t-test or Wilcoxon signed-rank test
- **RQ3:** Spearman correlation
- **Scale quality:** Cronbach's alpha

No numerical findings are stored in the repository until real data are collected.


```python
from pathlib import Path
import sys
import pandas as pd

ROOT = Path.cwd().resolve()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent

sys.path.insert(0, str(ROOT))

from src.analysis import (
    cronbach_alpha,
    spearman_test,
    friedman_distraction_test,
    paired_condition_test,
)

PROCESSED = ROOT / "data" / "processed"
RESULTS = ROOT / "results" / "tables"
RESULTS.mkdir(parents=True, exist_ok=True)
```

```python
q_path = PROCESSED / "questionnaire_clean.csv"

if q_path.exists():
    q = pd.read_csv(q_path)
    print("Questionnaire rows:", len(q))
    display(q.head())
else:
    q = None
    print("Run 01_data_preparation.ipynb after adding questionnaire.csv.")
```

```python
if q is not None:
    distraction_cols = [
        "messaging_freq",
        "notifications_freq",
        "browser_tab_switch_freq",
        "ide_browser_switch_freq",
        "background_media_freq",
        "unrelated_social_freq",
    ]

    descriptive = q[distraction_cols].describe().T[["mean", "std", "50%"]]
    display(descriptive)

    rq1 = friedman_distraction_test(q, distraction_cols)
    print("RQ1 Friedman:", rq1)
```

```python
if q is not None:
    cli_items = [
        "cl_mental_demand",
        "cl_effort",
        "cl_focus_difficulty",
        "cl_refocus_difficulty",
    ]
    pi_items = ["p_delay", "p_avoid", "p_replace", "p_unrelated_switch"]

    print("CLI Cronbach alpha:", cronbach_alpha(q[cli_items]))
    print("PI Cronbach alpha:", cronbach_alpha(q[pi_items]))
```

```python
if q is not None:
    print("RQ2 survey:", spearman_test(q["context_switch_count_per_hour"], q["cli"]))
    print("RQ3:", spearman_test(q["cli"], q["pi"]))
```

```python
c_path = PROCESSED / "controlled_task_clean.csv"

if c_path.exists():
    c = pd.read_csv(c_path)
    controlled_result = paired_condition_test(c)
    print("RQ2 controlled task:", controlled_result)
else:
    print("No controlled-task data yet.")
```
